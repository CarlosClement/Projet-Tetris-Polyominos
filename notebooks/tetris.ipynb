> **Creer un chemin d'un point a a un point b a l'aide de polyominos**

```python
def pavage_1d_v1(tailles, longueur):
    tailles.sort()
    coefficients = [0] * len(tailles)
    while longueur > 0:
        entered = False
        for i in reversed(range(len(tailles))):
            if longueur >= tailles[i]:
                longueur -= tailles[i]
                coefficients[i] += 1
                entered = True
                break
        if not entered: return None
    return coefficients
```

```python
pavage_1d_v1([6, 8], 30)  # Problématique
```

```python
def pavage_1d_v2(tailles, longueur):
    if longueur == 0: return []
    for i in range(len(tailles)):
        if longueur >= tailles[i] and tailles[i] > 0:
            r = pavage_1d_v2(tailles, longueur - tailles[i])
            if r is not None: return r+[i]
    return None
```
