```python
def generate_pol(matrix:list[list[int]]):
    return to_square(matrix)
        
def to_square(matrix):
    #On trouve le maximum
    max_line = len(matrix) #ICI on permet une matrice comme dans les tests plus bas, où il y a plus de lignes que de colonnes
    for line in matrix:
        if len(line) > max_line:
            max_line = len(line)
    res = []
    i=0
    for line in matrix:
        res.append(line)
        minus = max_line - len(res[i])
        if minus > 0 :
            res[i] += [0 for i in range(minus)]         
        i+=1

    if len(matrix) < max_line:
        missinglines = max_line - len(matrix)
        for i in range(missinglines):
            res.append([0 for i in range(max_line)])

    return res
            
def cool_print(pol):
    for line in pol:
        print(line)
```

```python
def rotation(pol):
    new_matrix = []
    num_ligne=0
    for i in reversed(range(len(pol))): #On parcourt à l'envers
        new_matrix.append([])
        for line in pol:
            new_matrix[num_ligne].append(line[i])
        num_ligne += 1

    return to_square(new_matrix)
```

```python
def test_polyomino():
    p = generate_pol([[0,1],[1,0,1]])
    cool_print(p)
    p_prime = rotation(p)
    print("sans rotation")
    cool_print(p)
    print("rotation 1")
    cool_print(rotation(p))
    print("rotation 2")
    cool_print(rotation(rotation(p)))
    
test_polyomino()
```
